```python
def autokey_encrypt(plaintext, key):
    # remove sapces and convert into upper for plaintext, if it is alpha
    plaintext = ''.join(c for c in plaintext.upper() if c.isalpha())
    # remove spaces and convert into upper for key, if it is alpja
    key = ''.join(c for c in key.upper() if c.isalpha())
    # make a key stream by adding key and plain text and tehn slicing by starting at 0 and goind till len of plain text
    keystream = (key + plaintext)[:len(plaintext)]

    # make a list to store sipher 
    cipher = []
    # make a zip of cipher and key
    for p,k in zip(plaintext, keystream):
        # convert the letters into integers and take mod 
        c = (ord(p) - 65 + ord(k) - 65) % 26
        # append back by converting into char
        cipher.append(chr(c + 65))
    # return by joinig with empty string to return a string
    return ''.join(cipher)
```

```python
def autokey_decrypt(ciphertext, key):
    # remove sapces and convert into upper for plaintext, if it is alpha
    ciphertext= ''.join(c for c in ciphertext.upper() if c.isalpha())
    # remove spaces and convert into upper for key, if it is alpha
    key= ''.join(c for c in key.upper() if c.isalpha())

    # make a list for plain text and make list of key to append into it the text we will decrypt
    plaintext= []
    keystream = list(key)

    # run an enumeration loop on cipher text it will give index and character from it
    for i,c in enumerate(ciphertext):
        # fetxh the corresponding letter from key stream using itertator (i)
        k = keystream[i]
        # subtracting the key letter from cipher letter first converting it into intergers 
        p = ( (ord(c) - 65) - (ord(k) - 65) ) % 26
        # converting it into char by adding 65
        p_char = chr(p + 65)
        # appending the decrpyted in plaintext
        plaintext.append(p_char)
        # append the decrpyted in keystream for autokey 
        keystream.append(p_char)
    return ''.join(plaintext)
```

```python
key = "QUEENLY"
message = "ATTACK AT DAWN"
enc = autokey_encrypt(message, key)
dec = autokey_decrypt(enc, key)

print("Plaintext :", message)
print("Key       :", key)
print("Encrypted :", enc)
print("Decrypted :", dec)
```

```python
def caesar_encrypt(plaintext, key):
    result = []
    for ch in plaintext:
        if ch.isalpha():
            base = 65 if ch.isupper() else 97
            c = ( ord(ch) - base + key ) % 26
            result.append ( chr(c + base) )
        else:
            result.append(ch)
    return ''.join(result)
```

```python
def caesar_decrypt(ceasertext, key):
    return (caesar_encrypt(ceasertext, -key))
```

```python
ceaser = caesar_encrypt("Hello, World!", 3)   
plain = caesar_decrypt(ceaser, 3)  
```

```python
print(ceaser)
print(plain)
```

```python
def vigenere(text, key, decrypt=False):
    key = ''.join(c for c in key.upper() if c.isalpha())
    result = []
    j = 0

    for ch in text:
        if ch.isalpha():
            base = 65 if ch.isupper() else 97
            shift = ord(key[j % len(key)]) - 65
            if decrypt:
                shift = -shift
            result.append(chr ( (ord(ch) - base + shift) % 26 + base) )
            j += 1
        else:
            result.append(ch)
    return ''.join(result)
```

```python
def vigenere_encrypt(text, key):
    return vigenere(text, key)

def vigenere_decrypt(text, key):
    return vigenere(text, key, decrypt=True)
```

```python
enc = vigenere_encrypt("ATTACK at dawn", "LEMON")
print(enc)                                  # LXFOPV EF RNHR
print(vigenere_decrypt(enc, "LEMON"))       # ATTACK AT DAWN
```

```python
def frequency_analysis(text):
    frequency = {}

    for char in text:
        if char.isalpha():
            if char in frequency:
                frequency[char] += 1
            else:
                frequency[char] = 1
    return frequency
```

```python
freq = frequency_analysis('helloworld, how are you')
print(freq)
```
