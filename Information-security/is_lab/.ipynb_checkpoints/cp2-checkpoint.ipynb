```python
#ceaser cypher
def encrypt_caesar(text, key):
    result = ""

    for char in text:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')

            encrypted = chr((ord(char) - base + key) % 26 + base)

            result += encrypted
        else:
            result += char

    return result


def decrypt_caesar(text, key):
    return encrypt_caesar(text, -key)

```

```python
def print_frequencies(text):
    text = text.upper()

    frequency = {}

    for char in text:
        if char.isalpha():
            if char in frequency:
                frequency[char] += 1
            else:
                frequency[char] = 1

    total = sum(frequency.values())

    print("\nLetter Frequency")
    print("----------------")

    for letter in "ABCDEFGHIJKLMNOPQRSTUVWXYZ":
        count = frequency.get(letter, 0)

        if total > 0:
            percentage = (count / total) * 100
        else:
            percentage = 0

        print(letter, ":", count, "=", round(percentage, 2), "%")
```

```python
#vigenere
def encrypt_vigenere(plaintext, key):
    result = ""
    key = key.upper()

    key_index = 0

    for char in plaintext:
        if char.isalpha():

            p = ord(char.upper()) - ord('A')
            k = ord(key[key_index % len(key)]) - ord('A')

            encrypted = chr((p + k) % 26 + ord('A'))

            result += encrypted

            key_index += 1

        else:
            result += char

    return result


def decrypt_vigenere(ciphertext, key):
    result = ""
    key = key.upper()

    key_index = 0

    for char in ciphertext:
        if char.isalpha():

            c = ord(char.upper()) - ord('A')
            k = ord(key[key_index % len(key)]) - ord('A')

            decrypted = chr((c - k) % 26 + ord('A'))

            result += decrypted

            key_index += 1

        else:
            result += char

    return result
```

```python
#autokey
def encrypt_autokey(plaintext, key):
    plaintext = ''.join(
        char.upper() for char in plaintext if char.isalpha()
    )

    key = ''.join(
        char.upper() for char in key if char.isalpha()
    )

    # Generate key stream
    keystream = key

    while len(keystream) < len(plaintext):
        keystream += plaintext

    # Make sure key stream is exactly the plaintext length
    keystream = keystream[:len(plaintext)]

    ciphertext = ""

    for i in range(len(plaintext)):

        p = ord(plaintext[i]) - ord('A')
        k = ord(keystream[i]) - ord('A')

        c = (p + k) % 26

        ciphertext += chr(c + ord('A'))

    return ciphertext, keystream


def decrypt_autokey(ciphertext, key):
    ciphertext = ''.join(
        char.upper() for char in ciphertext if char.isalpha()
    )

    key = ''.join(
        char.upper() for char in key if char.isalpha()
    )

    plaintext = ""

    for i in range(len(ciphertext)):

        # Initial keyword
        if i < len(key):
            k = ord(key[i]) - ord('A')

        # Previously recovered plaintext
        else:
            k = ord(plaintext[i - len(key)]) - ord('A')

        c = ord(ciphertext[i]) - ord('A')

        p = (c - k) % 26

        plaintext += chr(p + ord('A'))

    return plaintext

```
