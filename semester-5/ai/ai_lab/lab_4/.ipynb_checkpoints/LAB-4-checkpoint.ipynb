```python
from collections import deque
```

## (3.1) 8 puzzle problem

```python
#state definations
#inital state
start = (3,4,1,
         2,5,6,
         7,8,0)

#end goal
goal = (1,2,3,
        4,5,6,
        7,8,0)
         
```

```python
def print_board(state):
    for i in range (0,9,3):
        print(state[i:i+3])
    print()
```

```python
def get_neighbours(state):
    blank = state.index(0)
    row, col = divmod(blank, 3)

    moves = [
        ("Up", -1,0),
        ("Down", 1,0),
        ("Left", 0,-1),
        ("Right", 0,1)
    ]

    neighbours = []

    for name, dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col
            board = list(state)
            board[blank], board[new_blank] = (board[new_blank], board[blank])
            neighbours.append((name,tuple(board)))
    return neighbours
```

```python
def solve_puzzle(start, goal):
    queue = deque([start])
    # dictionay that has key as state , then stores its parent and move that reached to it in a tuple
    parent = {start: (None, None)}
    while queue:
        current = queue.popleft()
        if current == goal:
            break;
        for move, next_state in get_neighbours(current):
            if next_state not in parent:
                parent[next_state] = (current, move)
                queue.append(next_state)

    if goal not in parent:
        print("no solution found")
        return
    path = []
    current = goal
    while current is not None:
        previous, move = parent[current]
        path.append((move, current))
        current = previous
    path.reverse()

    for move, state in path:
        print("move: " ,move if move else "Start")
        print_board(state)
    print("total moves: ", len(path)-1)
    
```

```python
solve_puzzle(start,goal)
```

## (3.2) decision tree

```python
tree = {
    "q": "Is it fully electric?",
    "yes": {
        "q": "Is it a sedan (separate trunk)?",
        "yes": "Tesla Model 3",
        "no": "Nissan Leaf",
    },
    "no": {
        "q": "Is it a 2-seat sports car?",
        "yes": {"q": "Is it Italian?", "yes": "Ferrari 488", "no": "Porsche 911"},
        "no": {
            "q": "Does it have an open cargo bed?",
            "yes": "Ford F-150",
            "no": {
                "q": "Is it built for off-roading?",
                "yes": "Jeep Wrangler",
                "no": {
                    "q": "Is it a luxury car?",
                    "yes": "Mercedes S-Class",
                    "no": "Toyota Corolla",
                },
            },
        },
    },
}

```

```python
def ask(question):
    while True:
        answer = input(question + " (y/n): ").strip().lower()
        if answer in ("y", "n"):
            return answer == "y"
        print("Please type y or n.")
```

```python
def identify(node):
    while isinstance(node, dict):       
        node = node["yes"] if ask(node["q"]) else node["no"]
    print("I think it's a", node + "!")
```

```python
print("Think of a car: Tesla Model 3, Nissan Leaf, Ferrari 488, Porsche 911,")
print("Ford F-150, Jeep Wrangler, Mercedes S-Class or Toyota Corolla.")
identify(tree)
```

## (3.3) cannibals missonaries

```python
# state = (cannibals_left, missionaries_left, boat_side)   boat_side: 0 = left, 1 = right
initial_state = (3, 3, 0)
goal_state    = (0, 0, 1)
```

```python
# possible boat loads: (cannibals, missionaries)
MOVES = [(1, 0), (0, 1), (1, 1), (2, 0), (0, 2)]
```

```python
def is_valid(state):
    c, m, b = state
    rc = 3 - c
    rm = 3 - m
    if c < 0 or m < 0 or rc < 0 or rm < 0:
        return False
    if m > 0 and c > m:        # cannibals outnumber missionaries on left
        return False
    if rm > 0 and rc > rm:     # cannibals outnumber missionaries on right
        return False
    return True
```

```python
def get_neighbours(state):
    c, m, b = state
    result = []
    for dc, dm in MOVES:
        if b == 0:             # boat on left, people go to the right
            new_state = (c - dc, m - dm, 1)
            direction = "right"
        else:                  # boat on right, people come back to the left
            new_state = (c + dc, m + dm, 0)
            direction = "left"
        if is_valid(new_state):
            move = "move " + str(dc) + " cannibal(s) and " + str(dm) + " missionary(ies) to the " + direction
            result.append((move, new_state))
    return result
```

```python
def print_board(state):
    c, m, b = state
    left  = "C" * c + "M" * m
    right = "C" * (3 - c) + "M" * (3 - m)
    if b == 0:
        print("  [", left, "] boat ~~~~~~~~~ [", right, "]")
    else:
        print("  [", left, "] ~~~~~~~~~ boat [", right, "]")
```

```python
def solve_puzzle(start, goal):
    queue = [start]
    head = 0                               # index of the front of the queue
    parent = {start: (None, None)}

    while head < len(queue):
        current = queue[head]
        head += 1
        if current == goal:
            break
        for move, next_state in get_neighbours(current):
            if next_state not in parent:
                parent[next_state] = (current, move)
                queue.append(next_state)

    if goal not in parent:
        print("no solution found")
        return

    path = []
    current = goal
    while current is not None:
        previous, move = parent[current]
        path.append((move, current))
        current = previous
    path.reverse()

    for move, state in path:
        print("move:", move if move else "Start")
        print_board(state)
    print("total moves:", len(path) - 1)
```

```python
solve_puzzle(initial_state, goal_state)
```
