```python
flavors = ["vanilla", "chocolate", "strawberry", "pistacchio"]
sauces = ["caramel", "butterscotch", "chocolate"]

def sundaes():
    for flavor in flavors:
        for sauce in sauces:
            print(flavor + " ice cream sundae with " + sauce + " sauce")

sundaes()
```

```python

def triangle():
    limit = 10
    value = 1
    row = 1
    while row <= value:
        if limit == 0:
            break
        column = 1
        while column <= row:
            if column != row:
                print(value, ' ', sep = '', end = '') 
            else:
                print(value)
            value = value + 1 
            column = column + 1
        row = row + 1
        limit = limit - 1

triangle()
```

```python
def factorial(n):
    if n < 0:
        print("please enter a positive number")
    if n == 0:
        return 1
    return n * factorial(n-1)

factorial(0)
```

```python
def count_pattern(lst1, lst2):
    sequence_count = 0
    len_of_sequence = len(lst1)

    for i in range(len(lst2) - len_of_sequence + 1):
        for j in range(len_of_sequence):
            if lst1[j] != lst2[j + i]:
                break
        else:
            sequence_count += 1

    print(sequence_count)
```

```python
def calculator():
    num1 = int(input("Enter number 1"))
    num2 = int(input("Enter number 2"))

    symbol = input("Enter +,-,*,/")
    match symbol:
        case '+': 
             return num1 + num2
        case '-':
             return num1 - num2
        case '*':
            return num1 * num2
        case '/':
            if num2 == 0 :
                raise Exception("cannot divide by 0")
            return num1 / num2

calculator()
```
