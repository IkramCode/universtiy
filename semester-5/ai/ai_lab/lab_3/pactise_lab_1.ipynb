```python
text = "hi my name is Ikramulhaq"
split_text = text.split()
print(split_text)
```

```python
biggest_word = max(split_text, key=len)
print(biggest_word)
```

```python
newtext = text.replace(biggest_word, "hi")
print(newtext)
```
