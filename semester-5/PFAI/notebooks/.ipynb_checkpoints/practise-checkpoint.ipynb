## ITER

```python
numbers = [10,20]

itobj = iter(numbers)
print(itobj)
print(next(itobj))
print(next(itobj))
#print(next(itobj))
```

```python
t = (1, 2)
s = {3, 4}
l = [5, 6]
string = 'ikram'
d = {'x':1, 'y':5}
```

```python
for obj in (t, s, l, string, d):
    it = iter(obj)
    print(type(it).__name__, "->", next(it))
```

```python
# sentinel in iter is a value that tells it to run untill the specific value appears
# dice rolling untill a 6 appears
import random
def roll_dice():
    return random.randint(1,6)
```

```python
rolls = iter(roll_dice, 6)
print(list(rolls))
```

## GENERATOR

```python
def count_up(n):
    for i in range(1, n + 1):
        yield i

generator_obj = count_up(4)
print(generator_obj)

while True:
    try:
        print(next(generator_obj))
    except StopIteration:
        break
        
```

```python
def demo():
    x = 300
    print("start")
    yield x
    x *= 4
    print("resumed")
    yield x

gen_obj = demo()
print(next(gen_obj))
print(next(gen_obj))
```

```python
data = [1, 2, 3, 4]

gen_obj = (n*n for n in data if n % 2 == 0)
print(next(gen_obj))
print(next(gen_obj))

```

```python
def squares(n):
    for i in range(n):
        yield i*i
generator_obj = squares(100)

while True:
    try:
        print(next(generator_obj))
    except StopIteration:
        break
```

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    yield a
    
generator_obj = fibonacci(10)

while True:
    try:
        print(next(generator_obj))
    except StopIteration:
        break
```

```python
from itertools import islice

def natural():
    i = 1
    while True:
        yield i
        i = i + 1
        
generator_obj = natural()

first_five = list(islice(generator_obj, 5))

```

```python
first_five
```

```python
gen_exp = (n ** 3 for n in first_five)

gen_exp
while True:
    try:
        print(next(gen_exp))
    except StopIteration:
        break
```

## Closures

```python
def multiplier(n):
    def multiply(x):
        return x * n
    return multiply

with_two = multiplier(2)
with_three = multiplier(3)
```

```python
print(with_two(2))
print(with_three(2))
```

```python
print(multiplier)
print(with_two)
print(with_three)
```

```python
def make_counter(start, step):
    count = start
    def increase_count():
        nonlocal count
        count += step
        return count
    return increase_count

counter = make_counter(0,2)
print("first call:", counter())
print("second call:", counter())
print("thierd call:", counter())
```

## Decorators

```python
import time
def time_calc(func):
    def wrapper():
        s_time = time.time()
        func()
        e_time = time.time()
        print(f"{func.__name__} took {e_time - s_time: .8f} seconds")
    return wrapper
```

```python
@time_calc
def mig25():
    print("mig25 taking off")
    time.sleep(3)
    print("mig25 took off")
    
@time_calc
def f16():
    print("f16 taking off")
    time.sleep(1.5)
    print("f16 took off")

mig25()
f16()
```

```python
def string_upper(func):
    def wrapper():
        string = func()
        upper_str= string.upper()
        print("string into upper case :",upper_str)
    return wrapper
```

```python
@string_upper
def return_some_string(string):
    return string

return_some_string("plz work")
```
