```python
from functools import reduce
```

```python
print(list(map(lambda w:w.upper(),(['pyhotn','rust']))))
```

```python
nums = [1,2,3,4,5,6]
print(list(map(lambda num:num*2, nums)))
```

```python
def convert_to_farenhite(x):
    return x * 9/5 + 32
```

```python
print(list(map(convert_to_farenhite, [0,20,37,100])))
```

```python
print(list(map(lambda x,y:x*y,[1,2,3],[4,5,6])))
```

```python
print(list(filter(lambda x:x%2==0, [1,2,3,4,5,6,7,8])))
```

```python
words = ["python", "go", "rust", "javascript", "c"]
def string_greater_than_3(x):
    return(len(x) > 3)
```

```python
print(list(filter(string_greater_than_3, words)))
```

```python
def is_fist_char_upper(x):
    return(x[0].isupper())
```

```python
names = ["Ali", "ahmed", "Sara", "usman", "Zain"]
```

```python
list(filter(is_fist_char_upper, names))
```

```python
reduce(lambda acc,x: acc + x, nums, 100)
```

```python
reduce(lambda acc,x: acc * x, nums)
```

```python
reduce(lambda a,b: a if a > b else b , [3, 9, 2, 7])
```

```python
reduce(lambda acc,x: acc + (1 if x % 2 == 0 else 0), [1, 2, 3, 4], 0)
```

```python
reduce(lambda x,y: x + "-" + y,words)
```
