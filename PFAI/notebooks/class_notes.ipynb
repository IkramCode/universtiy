### pyhton builtin functions - ZIP

```python
# zip(*iterables, strict=false/true)
# zip iterates over several iterables in parallel

# iterable -> str, list, tuple, dict etc
# zip returns an iterator of tuples , the ith tuple contains the ith item from each iterable
# stop when shortest executable is exhausted like in below example b is shortest executable
# ZIP is a lazy iterator object, READING ASSIGMENT
```

```python
a = [1,2,3]
b = ['x', 'y']
data=zip(a,b, strict=False)
print(list(data))
```

```python
a = [1,2,3]
b = ['x', 'y']
data=zip(a,b, strict=True) #gives an error
print(list(data))
```
