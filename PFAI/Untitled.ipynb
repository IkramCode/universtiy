```python
words = ["python", "go", "rust", "javascript", "c"]
def string_greater_than_3(x):
    return(len(x) > 3)
```

```python
print(list(filter(string_greater_than_3, words)))
```

```python
def is_fist_char_upper(x):
    return(x[0].isupper())
```

```python
names = ["Ali", "ahmed", "Sara", "usman", "Zain"]
```

```python
list(filter(is_fist_char_upper, names))
```

```python
list(map(lambda x: x > 2, [1, 2, 3, 4]))     # [False, False, True, True]  (the results)
```

```python
list(filter(lambda x: x > 2, [1, 2, 3, 4]))  # [3, 4]  (the original items that passed)
```

```python

def f(x,y):
    return x,y;
```

```python
reduce(f, [5,6])         # 5 (f is never called with a single item!)
```

```python
reduce(lambda a, b: a - b, [10, 2, 3])   # 5, not 11: it's (10 - 2) - 3
```

```python
reduce(lambda a, b: b - a, [10, 2, 3])   # order of acc/x matters for non-commutative ops
```
