```python
import heapq as hq
```

```python
graph = {
    'S': {'A':2, 'B':5},
    'A': {'C':2, 'D':4},
    'B': {'D':1},
    'C': {'G':5},
    'D': {'G':2},
    'G': {}
}
print(graph)
```

```python
def ucs(start, goal):
    pq = [(0, start)]
    visited = set()
    parent={start: None}

    while pq:
        cost, current = hq.heappop(pq)
        if current in visited:
            continue
            
        visited.add(current)
        
        if current == goal:
            break

        for neighbour, edge_cost in graph[current].items():
            if neighbour not in visited:
                parent[neighbour] = current

                hq.heappush(pq, (cost + edge_cost, neighbour))
                
        if goal not in visited:
            return None

    path = []
    node = goal
    while node is not None:
        path.append(node)
        node = parent[goal]
    path.reverse()
    return path
        
```

```python
path = ucs('S','G')
print(path)
```
