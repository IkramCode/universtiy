```python
graph = {
        'A': {'B':2, 'C':1},
        'B': {'D':5},
        'C': {'E':7, 'F':6},
        'D': {'G':1, 'H':8},
        'E': {'I':2, 'J':1},
        'F': {'K':12, 'L':9},
        'G': {},
        'H': {},
        'I': {},
        'J': {},
        'K': {},
        'L': {}
        }
```

```python
def bfs(start, goal):
    queue = [start]
    
    explored = []
    parent = {start: None}
    g_cost = {start: 0}
    
    while queue:
        node = queue.pop(0)
        if node in explored:
            continue
        explored.append(node)
        if node == goal:
            break
        
        for child, cost in graph[node].items():
            if child not in parent:
                queue.append(child)
                parent[child] = node
                g_cost[child] = g_cost[node] + cost
    
    if goal not in explored:
        return None, None, explored

    path = []
    node = goal
    while node is not None:
        path.append(node)
        node = parent[node]
    path.reverse()

    return path, g_cost[goal], explored    
```

```python
path= bfs('A','J')
path
```
