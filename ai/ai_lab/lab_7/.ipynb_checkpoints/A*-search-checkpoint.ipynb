```python
import heapq
```

```python
graph = {
    'S': {'A':2, 'B':5},
    'A': {'C':2, 'D':4},
    'B': {'D':1},
    'C': {'G':5},
    'D': {'G':2},
    'G': {}
}
print(graph)
```

```python
h = {
    'S': 7,
    'A': 4,
    'B': 3,
    'C': 3,
    'D': 2,
    'G': 0
}
print(h)
```

```python
def a_star(start, goal):
    #       f(n)    g(n)  node
    pq = [(h[start], 0, start)]
    # list for explored nodes
    explored =[]
    # parent of each node to trackback
    parent = {start: None}
    # A* uses it
    g_cost = {start: 0}

    while pq:
        _, cost, current = heapq.heappop(pq)
        
        if current in explored:
            continue
            
        explored.append(current)
        
        if current == goal:
            break

        for neighbour, edgecost in graph[current].items():
            new_cost = g_cost[current] + edgecost

            if neighbour not in g_cost or new_cost < g_cost[neighbour]:
                g_cost[neighbour] = new_cost
                parent[neighbour] = current

                f_cost = new_cost + h[neighbour]
                heapq.heappush(pq, (f_cost, new_cost, neighbour))
                
    if goal not in explored:
        print("Goal not found")
        return

    path = []
    node = goal

    while node is not None:
        path.append(node)
        node = parent[node]
    path.reverse()

    print("Explored nodes",explored)
    print("final path",path)
    print("total g_cost:",g_cost[goal])
    
```

```python
a_star('S','G')
```

## 8 PUZZLE

```python
start = initial_state = (1, 2, 3,
                 4, 5, 6,
                 0, 7, 8) 

final_state =    (1,2,3,
                 4,5,6,
                 7,8,0)

def print_board(state):
    for i in range (0,9,3):
        print(state[i:i+3])
    print()

print_board(initial_state)
```

```python
def get_neighbours(state):
    neighbours = {}
    blank = state.index(0)
    row, col = divmod(blank, 3)

    moves = [(-1,0), (1,0), (0,-1), (0,1)]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col
            new_state = list(state)
            new_state[blank], new_state[new_blank] = new_state[new_blank], new_state[blank]
            neighbours[tuple(new_state)] = 1
    return neighbours
```

```python
def manhattan(state, goal):
    # sum of row + column distance of each tile from its goal position
    total = 0
    for i, tile in enumerate(state):
        if tile == 0:
            continue
        goal_index = goal.index(tile)
        r1, c1 = divmod(i, 3)
        r2, c2 = divmod(goal_index, 3)
        total += abs(r1 - r2) + abs(c1 - c2)
    return total
```

```python
print(manhattan(start, final_state))
```

```python
def a_star(start, goal):
    #       f(n)                 g(n)  node
    pq = [(manhattan(start,goal), 0, start)]
    # list for explored nodes
    explored =[]
    # parent of each node to trackback
    parent = {start: None}
    # A* uses it
    g_cost = {start: 0}

    while pq:
        _, cost, current = heapq.heappop(pq)
        
        if current in explored:
            continue
            
        explored.append(current)
        
        if current == goal:
            break

        for neighbour, edgecost in get_neighbours(current).items():
            new_cost = g_cost[current] + edgecost

            if neighbour not in g_cost or new_cost < g_cost[neighbour]:
                g_cost[neighbour] = new_cost
                parent[neighbour] = current

                f_cost = new_cost + manhattan(neighbour, goal)
                heapq.heappush(pq, (f_cost, new_cost, neighbour))
                
    if goal not in explored:
        print("Goal not found")
        return

    path = []
    node = goal

    while node is not None:
        path.append(node)
        node = parent[node]
    path.reverse()

    print("Explored nodes",explored)
    print("final path",path)
    print("total g_cost:",g_cost[goal])
    
```

```python
a_star(start, final_state)
```
