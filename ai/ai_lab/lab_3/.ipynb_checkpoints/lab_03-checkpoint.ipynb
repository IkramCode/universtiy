## initializing grpah

```python
graph = dict([
    ('A', ['B','C']),
    ('B', ['A','D','E']),
    ('C', ['A', 'F']),
    ('D', ['B']),
    ('E', ['B', 'G']),
    ('F', ['C']),
    ('G', ['E'])
])

graph
```

```python
def display_graph(graph):
    for node in graph:
        print(node, "->", graph[node])

display_graph(graph)
```

```python
def get_nodes(graph):
    return list(graph)

get_nodes(graph)
```

```python
def count_nodes(graph):
    return len(graph)

print(count_nodes(graph))
```

```python
def get_neighbours(graph, node):
    return(graph.get(node, []))

get_neighbours(graph, 'B')
```

```python
def node_exist(graph, node):
    return node in graph

print(node_exist(graph, 'B'))
```

```python
def add_node(graph, node):
    if node not in graph:
        graph[node] = []
        
add_node(graph, 'H')
get_nodes(graph)
```

```python
def add_edge(graph, node1, node2):
    if node1 not in graph:
        add_node(graph, node1)
    if node2 not in graph:
        add_node(graph, node2)
        
    if node2 not in graph[node1]:
        graph[node1].append(node2)
    
    if node1 not in graph[node2]:
        graph[node2].append(node1)

add_edge(graph, 'G', 'H')
display_graph(graph)
```

```python
def remove_edge(graph, node1, node2):
    if node1 in graph and node2 in graph[node1]:
        graph[node1].remove(node2)
    if node2 in graph and node1 in graph[node2]:
        graph[node2].remove(node1)

remove_edge(graph, 'G', 'H')
display_graph(graph)
```

```python
def remove_node(graph, node):
    for neighbour in graph[node]:
        graph[neighbour].remove(node)
    del graph[node]

remove_node(graph, 'H')
display_graph(graph)
```

```python
def get_edges(graph):
    edges = []
    for node in graph:
        for neighbor in graph[node]:
            if (neighbor, node) not in edges:
                edges.append((node, neighbor))
    return edges

print(get_edges(graph))
```

```python
def isolated_nodes(graph):
    return [node for node in graph if len(graph[node]) == 0]

print ("isolated nodes: ", isolated_nodes(graph))
```

## FIND PATH BETWEEN TWO NODES

```python
def find_path(graph, start, goal):
    if start not in graph or goal not in graph:
        return None

    stack = [[start]]

    while stack:
        path = stack.pop()
        node = path[-1]

        if node == goal:
            return path

        for neighbour in reversed(graph[node]):
            if neighbour not in path:
                stack.append(path + [neighbour])
                
    return None

print(find_path(graph, 'A','G'))
print(find_path(graph, 'A','H'))
```

## FIND ALL PATHS

```python
def find_all_paths(graph, start, goal):
    all_paths = []
    if start not in graph or goal not in graph:
        return None

    stack = [[start]]

    while stack:
        path = stack.pop()
        node = path[-1]

        if node == goal:
            all_paths.append(path)
            continue

        for neighbour in reversed(graph[node]):
            if neighbour not in path:        
                stack.append(path + [neighbour])
    return all_paths
```

```python
def find_all_paths_weighted(graph, start, goal):
    all_paths = []
    if start not in graph:
        return all_paths

    stack = [[start]]

    while stack:
        path = stack.pop()
        node = path[-1]

        if node == goal:
            all_paths.append(path)     
            continue                   

        for neighbor, weight in reversed(graph[node]):
            if neighbor not in path:   
                stack.append(path + [neighbor])
                
    return all_paths
```

```python
graph2 = dict([
    ('A', ['B', 'C']),
    ('B', ['A', 'D']),
    ('C', ['A', 'D', 'F']),
    ('D', ['B', 'C', 'E']),
    ('E', ['D', 'F']),
    ('F', ['C', 'E'])
])

paths = find_all_paths(graph2, 'A', 'F')

for i, p in enumerate(paths, 1):
    print(f"Path {i}:", p)

print("Total paths:", len(paths))
```

## Find shortest path

```python
def find_shortest_path(graph, start, end):
    paths = find_all_paths(graph2, start, end)

    if not paths:
        return None

    return min(paths, key = len)
```

## Find shortest wighted path

```python
def path_cost(graph, path):
    total = 0
    for i in range(len(path) - 1):
        for neighbor, weight in graph[path[i]]:
            if neighbor == path[i + 1]:
                total += weight
                break
    return total
```

```python
graph2_directed_weighted = dict([
    ('A', [('B', 4), ('C', 2)]),
    ('B', [('D', 5)]),
    ('C', [('D', 8), ('F', 10)]),
    ('D', [('E', 2)]),
    ('E', [('F', 3)]),
    ('F', [])
])
```

```python
def find_shortest_path_weighted(graph, start, end):
    paths = find_all_paths_weighted(graph, start, end)      # graph, not graph2

    if not paths:
        return None

    return min(paths, key=lambda p: path_cost(graph, p))   # weight, not len

```

```python
print('shortest path from A to F')
path = find_shortest_path_weighted(graph2_directed_weighted, 'A', 'F')
print(path)                                          # ['A', 'C', 'F']
print(path_cost(graph2_directed_weighted, path))     # 12
```

## Find longest weighted path

```python
def find_longest_path_weighted(graph, start, end):
    paths = find_all_paths_weighted(graph, start, end)

    if not paths:
        return None

    return max(paths, key=lambda p: path_cost(graph, p))
```

```python
print('longest path from A to F')
path = find_longest_path_weighted(graph2_directed_weighted, 'A', 'F')
print(path)                                          
print(path_cost(graph2_directed_weighted, path))     
```

## University campus navigation

```python
university_map_graph = dict([
    ('Main-gate',[('Reception', 4)]),
    ('Reception', [('Library', 6),('Cafe' ,3)]),
    ('Library' ,[('Lab', 2)]),
    ('Cafe' ,[('Department', 4)]),
    ('Lab' ,[('Department', 2), ('Parking', 2)]),
    ('Department' ,[('Parking', 5)]),
    ('Parking' ,[('Department', 5), ('Lab', 2)])
])
```

```python
all_main_to_lab_paths = find_all_paths_weighted(university_map_graph, 'Main-gate', 'Lab')
shortest_main_to_lab_paths = find_shortest_path_weighted(university_map_graph, 'Main-gate', 'Lab')

print('All paths:')
print(all_main_to_lab_paths)

for p in all_main_to_lab_paths:
    cost = path_cost(university_map_graph, p)
    print(" -> ".join(p), f"(distance: {cost})")

print('shortest path:')
print(shortest_main_to_lab_paths)

print(path_cost(university_map_graph, shortest_main_to_lab_paths))     
```

## Social Network

```python
social = dict([
    ('Ali', ['Ahmed', 'Sara']),
    ('Ahmed', ['Ali', 'Hassan']),
    ('Hassan', ['Ahmed', 'Fatima']),
    ('Sara', ['Ali', 'Fatima']),
    ('Fatima', ['Hassan', 'Sara'])
])
```

### Find all friends of a person

```python
person = 'Ali'
print(f"Friends of {person}:", get_neighbours(social, person))
```

### Check whether two people are directly connected

```python
def directly_connected(graph, p1, p2):
    return p2 in graph.get(p1, [])

print('Ali and Ahmed directly connected:', directly_connected(social, 'Ali', 'Ahmed'))
print('Ali and Hassan directly connected:', directly_connected(social, 'Ali', 'Hassan'))
```

### Find a path between two people

```python
path = find_path(social, 'Ali', 'Hassan')
print(" -> ".join(path))
```

### Find all possible connection paths

```python
all_connections = find_all_paths(social, 'Ali', 'Hassan')

for i, p in enumerate(all_connections, 1):
    print(f"Path {i}:", " -> ".join(p))

print("Total paths:", len(all_connections))
```

### Minimum number of connections between two people

```python
shortest_connection = min(all_connections, key=len)

print("Shortest connection:", " -> ".join(shortest_connection))
print("Minimum number of connections:", len(shortest_connection) - 1)
```
